```python
from ultralytics import YOLO

# ==========================================
# NOTE POUR L'UTILISATEUR
# ==========================================
# Ce script permet de ré-entraîner le modèle sur votre propre machine.
# Assurez-vous que le dossier 'dataset_final' est dans le même répertoire.
# ==========================================

def main():
    # 1. Charger un modèle de base (nano : le plus léger et rapide)
    print("Chargement du modèle pré-entraîné YOLOv8n...")
    model = YOLO('yolov8n.pt') 

    # 2. Lancer l'entraînement
    # Note : Sur Windows, il est crucial de garder ce code dans la fonction main()
    # et d'utiliser 'workers=0' pour éviter les erreurs de mémoire.
    print("Démarrage de l'entraînement...")
    
    results = model.train(
        data='dataset_final/data.yaml', # Chemin vers le fichier de config
        epochs=50,                      # Nombre de passages sur les données (50 est un bon début)
        imgsz=640,                      # Taille des images (standard YOLO)
        batch=8,                        # Réduisez à 4 ou 8 si vous avez peu de mémoire GPU/RAM
        workers=0,                      # OBLIGATOIRE sur Windows pour éviter "Kernel died"
        name='nouvel_entrainement',     # Nom du dossier de sauvegarde
        device=0                        # Mettre 'cpu' si vous n'avez pas de carte graphique NVIDIA
    )

    print("Entraînement terminé !")
    print(f"Le nouveau modèle est sauvegardé dans : runs/detect/nouvel_entrainement/weights/best.pt")

# Protection nécessaire pour Windows (Multiprocessing)
if __name__ == '__main__':
    main()
```
